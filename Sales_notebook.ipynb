```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("Files/Superstore_raw.csv")

df.show()
```

```python
# Clean column names (remove spaces & special chars)
new_columns = [col.replace(" ", "_").replace("-", "_") for col in df.columns]
df = df.toDF(*new_columns)

df.show()
```

```python
df.write.mode("overwrite").saveAsTable("sales_raw")
```

```python
from pyspark.sql.functions import to_date, col


df = df.withColumn("order_date", to_date(col("order_date"), "MM-dd-yyyy"))
df = df.withColumn("ship_date", to_date(col("ship_date"), "MM-dd-yyyy"))
```

```python
df.show()
```

```python
df = df.dropna()
```

```python
df=df.drop_duplicates()
```

```python
from pyspark.sql.functions import col

df = df.withColumn("sales", col("sales").cast("double"))
df = df.withColumn("profit", col("profit").cast("double"))
```

```python
df.write.mode("overwrite").saveAsTable("sales_clean")
```
